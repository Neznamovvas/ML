3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
high = df[df['Stress Levels'] > 7]
low  = df[df['Stress Levels'] <= 3]

sbp_high = high['Blood Pressure (s/d)'].str.split('/').str[0].astype(float)
sbp_low  = low['Blood Pressure (s/d)'].str.split('/').str[0].astype(float)
dbp_high = high['Blood Pressure (s/d)'].str.split('/').str[1].astype(float)
dbp_low  = low['Blood Pressure (s/d)'].str.split('/').str[1].astype(float)
eps = 5.0

sbp_diff = sbp_high.mean() - sbp_low.mean()
dbp_diff = dbp_high.mean() - dbp_low.mean()

print(f"SBP: высокий = {sbp_high.mean():.1f}, низкий = {sbp_low.mean():.1f}, "
      f"разница = {sbp_diff:+.1f}")
print(f"DBP: высокий = {dbp_high.mean():.1f}, низкий = {dbp_low.mean():.1f}, "
      f"разница = {dbp_diff:+.1f}")
print(f"Порог значимости eps = {eps} мм рт. ст.")

if sbp_diff > eps and dbp_diff > eps:
    print("Да: у людей с высоким стрессом кровяное давление в среднем выше.")
elif abs(sbp_diff) <= eps and abs(dbp_diff) <= eps:
    print("Нет: разница в давлении между группами в пределах eps (ничтожна).")
else:
    print("Нет: давление у группы высокого стресса не выше значимо.")
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
chol_m = df[df['Gender'] == 'Male']['Cholesterol Level (mg/dL)']
chol_f = df[df['Gender'] == 'Female']['Cholesterol Level (mg/dL)']

print(f"Мужчины:  n = {len(chol_m)}, среднее = {chol_m.mean():.2f}, медиана = {chol_m.median():.2f}")
print(f"Женщины:  n = {len(chol_f)}, среднее = {chol_f.mean():.2f}, медиана = {chol_f.median():.2f}")

mean_diff = chol_m.mean() - chol_f.mean()
median_diff = chol_m.median() - chol_f.median()

eps = 1.0

print(f"Разница средних: {mean_diff:+.2f}")
print(f"Разница медиан:  {median_diff:+.2f}")
print(f"Порог значимости eps = {eps} мг/дл")

if abs(mean_diff) > eps:
    print("Да: уровень холестерина у мужчин и женщин в среднем отличается.")
else:
    print("Нет: разница среднего уровня холестерина в пределах eps (ничтожна).")
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = df[df['Gender'] == 'Female']

young = women[women['Age (years)'] < 30]['Cognitive Function']
old   = women[women['Age (years)'] >= 60]['Cognitive Function']

print(f"Юные (< 30):      n = {len(young)}, среднее = {young.mean():.2f}, медиана = {young.median():.2f}")
print(f"В возрасте (60+):  n = {len(old)}, среднее = {old.mean():.2f}, медиана = {old.median():.2f}")

mean_diff = old.mean() - young.mean()
median_diff = old.median() - young.median()

eps = 1.5

print(f"Разница средних (60+ минус <30): {mean_diff:+.2f}")
print(f"Разница медиан (60+ минус <30):  {median_diff:+.2f}")
print(f"Порог значимости eps = {eps}")

if len(young) == 0 or len(old) == 0:
    print("Недостаточно данных в одной из групп.")
elif mean_diff < -eps:
    print("Да: с возрастом женщины оценивают когнитивные способности хуже, чем в юности.")
elif abs(mean_diff) <= eps:
    print("Нет: разница в оценке когнитивных способностей в пределах eps (ничтожна).")
else:
    print("Нет: женщины 60+ оценивают когнитивные способности не хуже, чем в юности.")
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
p30 = df['BMI'].quantile(0.30)
p60 = df['BMI'].quantile(0.60)
p75 = df['BMI'].quantile(0.75)

print(f"Пороги BMI: P30 = {p30:.2f}, P60 = {p60:.2f}, P75 = {p75:.2f}")

group_high = df[df['BMI'] > p75]
group_mid  = df[(df['BMI'] >= p30) & (df['BMI'] <= p60)]

high_rate = group_high['Chronic Diseases'].notna().mean()
mid_rate  = group_mid['Chronic Diseases'].notna().mean()

print(f"BMI > P75:   n = {len(group_high)}, доля с хрон. заболеваниями = {high_rate:.2%}")
print(f"BMI P30–P60: n = {len(group_mid)},  доля с хрон. заболеваниями = {mid_rate:.2%}")

diff = high_rate - mid_rate
eps = 0.01

print(f"Разница долей (high − mid): {diff:+.2%}, eps = {eps:.2%}")

if len(group_high) == 0 or len(group_mid) == 0:
    print("Недостаточно данных в одной из групп.")
elif abs(diff) <= eps:
    print("Нет: разница долей в пределах eps (ничтожна).")
elif diff > 0:
    print("Да: у людей с BMI > P75 хронические заболевания встречаются чаще.")
else:
    print("Нет: у людей с BMI > P75 хронические заболевания встречаются реже.")
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
import pandas as pd

df = pd.read_csv('data.csv')

high_income = df[df['Income Level'] == 'High']['Physical Activity Level']
low_income  = df[df['Income Level'] == 'Low']['Physical Activity Level']

mapping = {'Low': 1, 'Moderate': 2, 'High': 3}

high_score = high_income.map(mapping)
low_score  = low_income.map(mapping)

print(f"Высокий доход: n = {len(high_score)}, средний балл активности = {high_score.mean():.2f}")
print(f"Низкий доход:  n = {len(low_score)},  средний балл активности = {low_score.mean():.2f}")

diff = high_score.mean() - low_score.mean()
eps = 0.1

print(f"Разница (high − low): {diff:+.2f}, eps = {eps}")

if len(high_score) == 0 or len(low_score) == 0:
    print("Недостаточно данных в одной из групп.")
elif abs(diff) <= eps:
    print("Нет: разница в уровне активности в пределах eps (ничтожна).")
elif diff > 0:
    print("Да: у людей с высоким доходом уровень физической активности выше.")
else:
    print("Нет: у людей с высоким доходом уровень физической активности ниже.")
````

2. Как связана оценка собственных когнитивных функций с уровнем стресса?

````python
import numpy as np

x = df['Stress Levels']
y = df['Cognitive Function']

corr = x.corr(y)
print(f"Корреляция Пирсона (стресс vs когнитивные функции): {corr:+.3f}")

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(x, y, alpha=0.3, s=15, color='steelblue')

k, b = np.polyfit(x, y, 1)
xs = np.linspace(x.min(), x.max(), 100)
ax.plot(xs, k * xs + b, color='red', linewidth=2,
        label=f'тренд: y = {k:+.2f}x + {b:.2f}\nr = {corr:+.3f}')

ax.set_title('Связь уровня стресса и когнитивных функций')
ax.set_xlabel('Уровень стресса')
ax.set_ylabel('Оценка когнитивных функций')
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

6. Имеет ли связь зрение и способность слышать звуки на высоких частотах? Ответьте на вопрос - какая зависимость между ними и предположите почему зависимость именно такая?

````python
vision = df['Vision Sharpness']
hearing = df['Hearing Ability (dB)']
age = df['Age (years)']

mask = vision.notna() & hearing.notna() & age.notna()
v, h, a = vision[mask], hearing[mask], age[mask]

def residuals(y, x):
    k, b = np.polyfit(x, y, 1)
    return y - (k * x + b)

r_pair = v.corr(h)

rv = residuals(v, a)
rh = residuals(h, a)
r_partial = np.corrcoef(rv, rh)[0, 1]

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

sc = axes[0].scatter(v, h, c=a, cmap='viridis', alpha=0.6, s=14)
cb = plt.colorbar(sc, ax=axes[0])
cb.set_label('Возраст')

k, b = np.polyfit(v, h, 1)
xs = np.linspace(v.min(), v.max(), 100)
axes[0].plot(xs, k * xs + b, color='red', linewidth=2)

axes[0].set_title(f'Зрение и слух (r = {r_pair:.2f})')
axes[0].set_xlabel('Vision Sharpness')
axes[0].set_ylabel('Hearing Ability, dB')
axes[0].grid(alpha=0.3)

axes[1].scatter(rv, rh, alpha=0.4, s=14, color='slategray')

k, b = np.polyfit(rv, rh, 1)
xs = np.linspace(rv.min(), rv.max(), 100)
axes[1].plot(xs, k * xs + b, color='red', linewidth=2)

axes[1].axhline(0, color='black', linewidth=0.8, alpha=0.5)
axes[1].axvline(0, color='black', linewidth=0.8, alpha=0.5)
axes[1].set_title(f'С учётом возраста (r = {r_partial:.2f})')
axes[1].set_xlabel('Vision Sharpness')
axes[1].set_ylabel('Hearing Ability, dB')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"n = {len(v)}")
print(f"Парная корреляция:        r = {r_pair:+.3f}")
print(f"Частная (контроль возр.): r = {r_partial:+.3f}")
print(f"Падение |r|:              {abs(r_pair) - abs(r_partial):+.3f}")
````

Ответ:
В исходных данных — умеренная отрицательная корреляция (r = −0.639). С учётом инверсии шкалы Hearing Ability (dB) (меньше дБ = лучше слух) это означает положительную по смыслу связь: люди с лучшим зрением в среднем имеют лучший слух на высоких частотах.

Эта связь не является прямой и объясняется возрастом. После учёта возраста частная корреляция падает до r = +0.006 — практически до нуля. Это означает: с возрастом одновременно деградируют обе сенсорные системы (зрение — пресбиопия/катаракта; слух — пресбиакузис, возрастная потеря высоких частот). Общий фактор возраста и создаёт наблюдаемую корреляцию.
